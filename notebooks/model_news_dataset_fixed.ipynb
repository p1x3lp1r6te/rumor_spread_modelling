```python
# Save as train_three_lstm_models.py and run in an environment with:
# pandas, numpy, matplotlib, scikit-learn, tensorflow (Keras)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, InputLayer
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import tensorflow as tf
import math
import os

# ======== Parameters ========
CSV_PATH = "news_dataset_fixed.csv"   # uses path from your session
WINDOW_SIZE = 30
TEST_SPLIT = 0.30         # 65% train, 35% test (same as yours)
BATCH_SIZE = 32
EPOCHS = 100
PATIENCE = 10             # early stopping patience
LSTM_UNITS = 32           # smaller to avoid overfitting
L2_REG = 1e-4
DROPOUT = 0.2
RECURRENT_DROPOUT = 0.1
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)
# ===========================

# ---- 1) Load and preprocess ----
df = pd.read_csv(CSV_PATH)
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date').reset_index(drop=True)
df_index = df['date']  # keep for plotting
df = df[['total_news', 'rumor_news', 'non_rumor_news']]

values = df.values.astype('float32')

# We'll create sequences but fit scaler only on training rows to avoid leakage.
def create_sequences_from_scaled(scaled_arr, window):
    X, y = [], []
    for i in range(len(scaled_arr) - window):
        X.append(scaled_arr[i:i+window])         # window x features (3)
        y.append(scaled_arr[i+window])           # full 3 values (we'll select target later)
    return np.array(X), np.array(y)             # X: (N, window, 3), y: (N, 3)

# First do a conservative index split on raw values (so we know which rows belong to train)
num_rows = len(values)
split_idx = int((1 - TEST_SPLIT) * num_rows)

# Fit scaler on training *rows only* (not sequences)
scaler = MinMaxScaler()
scaler.fit(values[:split_idx])  # fit on training timeframe only
scaled_all = scaler.transform(values)

# Create sequences for all data (using scaled values).
X_all, y_all = create_sequences_from_scaled(scaled_all, WINDOW_SIZE)
# y_all are scaled triplets [total, rumor, non_rumor] at t+1 relative to the window

# Now compute sample split index for sequences:
# If sequence i uses rows i..i+window-1 as inputs and row i+window as output, the last index used for train is split_idx - 1
# So we want sequences where i+window <= split_idx - 1  -> i <= split_idx - 1 - window
seq_split = split_idx - WINDOW_SIZE
if seq_split <= 0:
    raise ValueError("Not enough train rows for chosen WINDOW_SIZE. Reduce WINDOW_SIZE or use more data.")

X_train, X_test = X_all[:seq_split], X_all[seq_split:]
y_train, y_test = y_all[:seq_split], y_all[seq_split:]

print("Shapes: X_train", X_train.shape, "X_test", X_test.shape, "y_train", y_train.shape, "y_test", y_test.shape)

# For each of the three targets, we'll build a separate model and y_target is the corresponding column from y_train/test
target_cols = ['total_news', 'rumor_news', 'non_rumor_news']
target_indices = [0, 1, 2]

# Helper: build model factory
def build_lstm_model(input_shape):
    model = Sequential([
        InputLayer(input_shape=input_shape),
        LSTM(LSTM_UNITS, kernel_regularizer=l2(L2_REG), recurrent_dropout=RECURRENT_DROPOUT),
        Dropout(DROPOUT),
        Dense(16, activation='relu', kernel_regularizer=l2(L2_REG)),
        Dense(1, activation='linear')
    ])
    model.compile(optimizer='adam', loss='mse')
    return model

# Callbacks
es = EarlyStopping(monitor='val_loss', patience=PATIENCE, restore_best_weights=True, verbose=1)
rlr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=max(3, PATIENCE//3), min_lr=1e-6, verbose=1)

# Storage
models = {}
histories = {}
results = {}
preds_test_scaled = {}

# Train one model per target
for col, idx in zip(target_cols, target_indices):
    print("\n\n=== Training model for target:", col, "===\n")
    y_tr = y_train[:, idx]  # scaled values
    y_te = y_test[:, idx]
    # y_tr,y_te shape (num_samples,)
    model = build_lstm_model((WINDOW_SIZE, 3))
    history = model.fit(
        X_train, y_tr,
        validation_data=(X_test, y_te),
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        callbacks=[es, rlr],
        verbose=2
    )
    models[col] = model
    histories[col] = history
    # predictions (scaled)
    y_pred_scaled = model.predict(X_test, verbose=0).reshape(-1)
    preds_test_scaled[col] = y_pred_scaled

    # plot loss
    plt.figure(figsize=(8,4))
    plt.plot(history.history['loss'], label='train_loss')
    plt.plot(history.history['val_loss'], label='val_loss')
    plt.title(f"Loss ({col})")
    plt.xlabel("Epoch")
    plt.ylabel("MSE")
    plt.legend()
    plt.grid(True)
    plt.show()

# Build a scaled array of predicted triplets for test rows:
# For each test timestep j, we have predicted scaled values for each col in preds_test_scaled.
preds_scaled_matrix = np.stack([preds_test_scaled['total_news'],
                                preds_test_scaled['rumor_news'],
                                preds_test_scaled['non_rumor_news']], axis=1)  # shape (N_test, 3)

# Inverse transform predictions and true y_test (which are scaled triplets)
y_test_unscaled = scaler.inverse_transform(y_test)            # true (N_test, 3)
preds_unscaled = scaler.inverse_transform(preds_scaled_matrix)  # (N_test, 3)

# Evaluate each model with unscaled metrics
def rmse(a,b): return math.sqrt(mean_squared_error(a,b))

for i, col in enumerate(target_cols):
    y_true = y_test_unscaled[:, i]
    y_pred = preds_unscaled[:, i]
    mae = mean_absolute_error(y_true, y_pred)
    r_mse = rmse(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    results[col] = {'MAE': mae, 'RMSE': r_mse, 'R2': r2}
    print(f"\nMetrics for {col}: MAE={mae:.3f}, RMSE={r_mse:.3f}, R2={r2:.3f}")

# ===== Check whether total_predicted == rumor_predicted + non_rumor_predicted =====
total_pred = preds_unscaled[:, 0]
rumor_pred = preds_unscaled[:, 1]
nonrumor_pred = preds_unscaled[:, 2]

# Raw difference
diff = total_pred - (rumor_pred + nonrumor_pred)
print("\nSum relation (pred_total - (pred_rumor + pred_nonrumor)) statistics:")
print(" mean difference:", np.mean(diff))
print(" std difference :", np.std(diff))
print(" max abs diff   :", np.max(np.abs(diff)))
print(" min diff       :", np.min(diff))

# Optionally evaluate equality after rounding to nearest integer
diff_rounded = np.round(total_pred) - (np.round(rumor_pred) + np.round(nonrumor_pred))
pct_exact_equal = np.mean(diff_rounded == 0) * 100.0
print(f"After rounding to integers, % exact equality = {pct_exact_equal:.2f}%")

# If you want to force consistency: simplest post-hoc fix is to set total = rumor + nonrumor
preds_unscaled_consistent = preds_unscaled.copy()
preds_unscaled_consistent[:, 0] = preds_unscaled_consistent[:, 1] + preds_unscaled_consistent[:, 2]

# Compare original predicted totals vs adjusted totals (small change if differences were tiny)
adj_diff = preds_unscaled_consistent[:, 0] - total_pred
print("Post-hoc adjust: mean change to total:", np.mean(adj_diff), "max change:", np.max(np.abs(adj_diff)))

# ---- Replace Block 4 with this timestamp-aware plotting ----
import matplotlib.dates as mdates

# Build the sequence target dates (date corresponding to y at i+WINDOW_SIZE)
# df_index was stored earlier: it's the full date series (aligned with `values`)
seq_target_dates = df_index[WINDOW_SIZE:].reset_index(drop=True)   # length == len(y_all)
test_dates = seq_target_dates[seq_split:].reset_index(drop=True)   # dates corresponding to y_test

# Choose how many test points to plot
Nplot = min(200, len(y_test_unscaled))
plot_dates = test_dates[:Nplot]

# Create the 3-subplot figure with timestamp x-axis
fig, axes = plt.subplots(3, 1, figsize=(16, 10), sharex=True)

for i, col in enumerate(target_cols):
    axes[i].plot(plot_dates, y_test_unscaled[:Nplot, i], label=f"{col} (true)", linewidth=1.5)
    axes[i].plot(plot_dates, preds_unscaled[:Nplot, i], label=f"{col} (pred)", linestyle='--', alpha=0.9)
    axes[i].set_ylabel("Count")
    axes[i].legend()
    axes[i].grid(True)

# X-axis formatting (nice adaptive date ticks)
axes[-1].set_xlabel("Date (timestamp)")
locator = mdates.AutoDateLocator(minticks=6, maxticks=12)
formatter = mdates.ConciseDateFormatter(locator)
axes[-1].xaxis.set_major_locator(locator)
axes[-1].xaxis.set_major_formatter(formatter)
plt.xticks(rotation=25)
plt.suptitle("True vs Predicted (test set) — timestamps on x-axis", fontsize=14)
plt.tight_layout(rect=[0, 0.03, 1, 0.96])
plt.show()


# Save models if you want
os.makedirs("trained_models", exist_ok=True)
for col, model in models.items():
    model.save(f"trained_models/lstm_{col}.keras")
print("\nSaved models to ./trained_models/")

# Done. results dict contains evaluation metrics per target.
print("\nSummary results:", results)
```
